```python
import re, statistics as st

rows = []
for s in [1, 2, 3]:
    txt = open(f"{REPO_DIR}/logs/tgn_wiki_seed{s}.txt").read()
    test = re.findall(r"Test:\s*mrr:\s*([0-9.]+)", txt)
    vals = [float(v) for v in re.findall(r"Validation mrr:\s*([0-9.]+)", txt)]
    epochs = len(re.findall(r"Epoch:\s*\d+", txt))
    tv = re.findall(r"Train & Validation: Elapsed Time \(s\):\s*([0-9.]+)", txt)
    if not test:
        print(f"seed {s}: no final test MRR in the log, it did not finish")
        continue
    rows.append((s, epochs, max(vals), float(test[0]), float(tv[0]) if tv else None))
    print(f"seed {s}: epochs run={epochs}, best val={max(vals):.4f}, test={float(test[0]):.4f}, train+val time={tv[0] if tv else '?'} s")

tests = [r[3] for r in rows]
if len(tests) >= 2:
    print(f"\nTest MRR mean ± std (sample std, n={len(tests)}): {st.mean(tests):.4f} ± {st.stdev(tests):.4f}")
    print("Published (TGB paper, 5 runs): 0.396 ± 0.060")

with open(f"{REPO_DIR}/results/tgn_wiki.csv", "w") as f:
    f.write("seed,epochs_run,best_val_mrr,test_mrr,train_val_time_s\n")
    for r in rows:
        f.write(",".join(str(x) for x in r) + "\n")
save_and_push("Add per-seed TGN tgbl-wiki summary CSV")
```

```python
import re, statistics as st

rows = []
for s in [1, 2, 3]:
    txt = open(f"{REPO_DIR}/logs/dyrep_wiki_seed{s}.txt").read()
    test = re.findall(r"Test:\s*mrr:\s*([0-9.]+)", txt)
    vals = [float(v) for v in re.findall(r"Validation mrr:\s*([0-9.]+)", txt)]
    tv = re.findall(r"Train & Validation[^\n]*?Elapsed Time \(s\):\s*([0-9.]+)", txt)
    if not test:
        print(f"seed {s}: no final test MRR, it did not finish")
        continue
    best_ep = vals.index(max(vals)) + 1
    rows.append((s, len(vals), best_ep, max(vals), float(test[0]), float(tv[0]) if tv else None))
    print(f"seed {s}: epochs={len(vals)}, best epoch={best_ep}, best val={max(vals):.4f}, test={float(test[0]):.4f}")

tests = [r[4] for r in rows]
vals_best = [r[3] for r in rows]
print(f"\nTest MRR mean ± std (n={len(tests)}): {st.mean(tests):.4f} ± {st.stdev(tests):.4f}")
print(f"Best val MRR mean ± std: {st.mean(vals_best):.4f} ± {st.stdev(vals_best):.4f}")
print("Published (TGB paper): val 0.072 ± 0.009, test 0.050 ± 0.017")

with open(f"{REPO_DIR}/results/dyrep_wiki.csv", "w") as f:
    f.write("seed,epochs_run,best_epoch,best_val_mrr,test_mrr,train_val_time_s\n")
    for r in rows:
        f.write(",".join(str(x) for x in r) + "\n")
save_and_push("Add per-seed DyRep tgbl-wiki summary CSV")
```
